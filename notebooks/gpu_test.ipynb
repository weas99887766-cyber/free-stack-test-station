```
import json, time, torch
assert torch.cuda.is_available(), 'GPU not detected'
a = torch.randn(2048, 2048, device='cuda'); b = torch.randn(2048, 2048, device='cuda')
torch.cuda.synchronize(); t0 = time.time(); _ = a @ b; torch.cuda.synchronize()
res = {'gpu': [torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())], 'matmul_ms': round((time.time()-t0)*1000, 1), 'status': 'gpu_ok'}
open('result.json', 'w').write(json.dumps(res, indent=2)); print(res)
```
